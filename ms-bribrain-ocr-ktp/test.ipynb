```python
import re
text = "P.SURABAYA ,270c-1996"

        # cari teks kapital
def abc(text):
    pattern = r"([A-Z]{2,}.*)"
    match = re.search(pattern, text)
    if not match:
        print("Tidak ada match")
        return "", ""

    content = match.group(1)
    print("match:", content)

    content = re.split(r'[,.]', content)
    print("split:", content)
    digit_data = len(re.findall(r"\d", str(content[1])))
    print(len(content[0]), digit_data)
    if len(content[0]) >= 3 and digit_data >= 5:
        number = re.sub(r'[^a-zA-Z0-9]', '', content[1])
        caps = content[0]
    else:
        # ekstrak tempat
        caps = re.findall(r"[A-Z][A-Z\.\- ]+", content)
        print("caps:", caps)
        # tempat = max(caps, key=len) if caps else ""
        tempat = tempat.replace(".", "").replace("-", " ").strip()

        # ekstrak angka
        numbers = re.findall(r"\d+", content)
        number = "".join(numbers)
        print("numbers:", numbers, "=>", number)
    if not number.isdigit():
        print("numbers:", "=>", number)
abc(text)
```
